## 9. Chained Conditionals — `28:10`

Combine boolean expressions with `and`, `or`, and `not`.

```python
x, y, z = 5, 7, 3

result_1 = x == y
result_2 = y > x
result_3 = z < x + 2

print(result_1, result_2, result_3)
print(result_1 or result_2)
print(result_2 and result_3)
print(not result_1)
print((y > x) and (z < x + 2))
```

## 11. Lists & Tuples — `35:41`

Lists are mutable ordered collections. Tuples are ordered but immutable.

```python
items = [4, True, "hello"]
items.append("new")
items.extend([8, 9])
print(items)

removed = items.pop(0)
print("removed:", removed)
print("first item:", items[0])

items[0] = "changed"
print(items)

coordinates = (52.52, 13.405)
print(coordinates[0])
# coordinates[0] = 1  # TypeError: tuples are immutable
```

### Mutability and copying

Assigning one list variable to another creates another reference to the same list. Use `.copy()` or slicing for a shallow copy.

```python
x = [1, 2, 3]
y = x
x[0] = 99
print("x:", x)
print("y:", y)

x = [1, 2, 3]
y = x.copy()
x[0] = 99
print("x copy example:", x)
print("y copy example:", y)
```

## 12. For Loops — `41:32`

A `for` loop iterates over an iterable such as a range, string, list, tuple, set, or dictionary.

## 14. Slicing — `47:33`

General form: `sequence[start:stop:step]`. The `stop` index is excluded.

```python
numbers = list(range(9))

print(numbers[2:5])
print(numbers[2:])
print(numbers[:4])
print(numbers[::2])
print(numbers[4:2:-1])
print(numbers[::-1])

text = "hello"
print(text[::2])
print(text[::-1])
```

## 15. Sets — `50:46`

Sets store unique values and are useful when membership matters more than order or duplicate frequency.

```python
values = {4, 3, 2, 2, 2}
print(values)

empty_set = set()
print(type(empty_set))
print(type({}))  # empty braces create a dict

values.add(10)
values.discard(3)
print(values)
print(10 in values)

left = {1, 2, 3}
right = {3, 4, 5}
print("union:", left | right)
print("intersection:", left & right)
```

## 16. Dictionaries — `54:29`

Dictionaries map unique keys to values.

```python
person = {
    "name": "Abed",
    "city": "Berlin",
    "skills": ["Python", "SQL"],
}

print(person["name"])
print(person.get("country", "Unknown"))

person["language"] = "Persian"
person["city"] = "Berlin, Germany"

for key, value in person.items():
    print(key, "->", value)
```

## 17. Comprehensions — `57:40`

Comprehensions create collections concisely from iterable data.

```python
squares = [i ** 2 for i in range(10)]
print(squares)

multiples_of_five = [i for i in range(100) if i % 5 == 0]
print(multiples_of_five)

mapping = {i: i ** 2 for i in range(5)}
print(mapping)

unique_squares = {i ** 2 for i in range(5)}
print(unique_squares)

as_tuple = tuple(i for i in range(5))
print(as_tuple)
```

## 18. Functions — `1:00:33`

Define functions with `def`. Parameters receive inputs; `return` sends a result back to the caller.

```python
def multiply(x, y):
    return x * y

print(multiply(5, 6))


def multiply_and_divide(x, y):
    return x * y, x / y

product, quotient = multiply_and_divide(5, 6)
print(product, quotient)


def greet(name, greeting="Hello"):
    return f"{greeting}, {name}!"

print(greet("Abed"))
print(greet("Abed", greeting="Hi"))
```

## 19. `*args`, `**kwargs` & Unpacking — `1:03:12`

- `*args` gathers extra positional arguments into a tuple.\n- `**kwargs` gathers extra keyword arguments into a dictionary.\n- `*` and `**` can also unpack collections when calling functions.

```python
def inspect_arguments(*args, **kwargs):
    print("args:", args)
    print("kwargs:", kwargs)

inspect_arguments(1, 2, 3, language="Python", level="fast")

numbers = [4, 5]
print(multiply(*numbers))

options = {"name": "Abed", "greeting": "Welcome"}
print(greet(**options))
```

### Functions are objects

Python functions can be stored in variables, passed around, and returned from other functions.

```python
def outer(x):
    def inner():
        return x
    return inner

returned_function = outer(42)
print(returned_function())
```

## 20. Scope & `global` — `1:09:12`

Names defined inside a function are local by default. Reading a global is allowed, but rebinding one requires `global`. In real projects, prefer returning values over changing global state.

```python
message = "global value"

def show_message():
    local_value = "local value"
    print(message)
    print(local_value)

show_message()

counter = 0

def increment_global():
    global counter
    counter += 1

increment_global()
print(counter)
```

## 21. Exceptions — `1:10:42`

Exceptions represent runtime errors. Instead of letting an expected error crash the program, you can handle it intentionally.

```python
try:
    int("not-a-number")
except ValueError as error:
    print("Caught:", type(error).__name__)
```

## 22. Handling Exceptions — `1:11:28`

Catch specific exceptions whenever possible. `else` runs if no exception occurs; `finally` always runs.

```python
def safe_divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        return None
    else:
        return result
    finally:
        print("division attempt finished")

print(safe_divide(10, 2))
print(safe_divide(10, 0))
```

## 23. Lambda Functions — `1:13:08`

A lambda is a small anonymous function containing a single expression.

```python
square = lambda x: x ** 2
add = lambda x, y: x + y

print(square(5))
print(add(3, 4))
```

## 24. `map()` & `filter()` — `1:14:19`

`map()` transforms each element. `filter()` keeps elements for which the function returns a truthy value. In modern Python, comprehensions are often more readable, but both functions are useful to recognize.

```python
numbers = [1, 2, 3, 4, 5, 6]

doubled = list(map(lambda i: i * 2, numbers))
even = list(filter(lambda i: i % 2 == 0, numbers))

print(doubled)
print(even)

# Equivalent comprehensions
print([i * 2 for i in numbers])
print([i for i in numbers if i % 2 == 0])
```

## 25. F-Strings — `1:17:30`

Prefix a string with `f` and place expressions inside `{}`.

```python
name = "Abed"
completed = 25
x = 6
y = 8

print(f"Hello, {name}!")
print(f"You completed {completed} sections.")
print(f"{x} + {y} = {x + y}")
print(f"Pi to 2 decimals: {3.14159265:.2f}")
```
